```python
# 1
def celsius_to_fahrenheit(celsius_array):
    
    return celsius_array * 9 / 5 + 32

# overall Test temperature change
test_c = np.array([0, 20, 37, 100])
print(celsius_to_fahrenheit(test_c))  
```

```python
# 3
def reshape_to_days(readings):
    
    return readings.reshape(7, 24)

# testing the results for reshape sensor
test_readings = np.arange(168)
result = reshape_to_days(test_readings)
print("shape:", result.shape)   
print("day 1:", result[0])
```

```python
# 4
def daily_min_max(daily_readings):
   
    return daily_readings.min(axis=1), daily_readings.max(axis=1)

# Test result for min/max temperature daily
test_data = reshape_to_days(np.arange(168))
mins, maxs = daily_min_max(test_data)
print("daily mins:", mins) 
print("daily maxs:", maxs)  
```

```python
# 5
# part 2
# total revenue by product 
def total_revenue_by_product(units_sold, unit_price):
    
    return units_sold.sum(axis=0) * unit_price

# Test results
test_units = np.array([[10, 5], [8, 6], [12, 4]])  
test_price = np.array([1000, 2000])
print(total_revenue_by_product(test_units, test_price))  
```

```python
# 6
# grade book weighed average
def weighted_final_scores(scores, weights):
    
    return scores @ weights

# Test results
test_scores = np.array([[90, 80, 70], [60, 70, 80], [100, 90, 95]])
test_weights = np.array([0.5, 0.3, 0.2])
print(weighted_final_scores(test_scores, test_weights)) 
```

```python
# 7
# solve for ingredient cost
def solve_ingredient_costs():
    

    A = np.array([[2, 1],
                  [1, 3]])
    b = np.array([3000, 4000])
    return np.linalg.solve(A, b)

print(solve_ingredient_costs())  
```

```python
# part c
# 8
# coin flip simulation
def simulate_coin_flips(n_flips=50000, seed=42):

    np.random.seed(seed)
    flips = np.random.randint(0, 2, size=n_flips)
    return flips.mean()

print(simulate_coin_flips())
```

```python
# 9
# customer wait time simulation
def pct_waiting_over_7min(n_customers=10000, seed=42):
    
    np.random.seed(seed)
    waits = np.random.normal(5, 1.5, size=n_customers)
    return (waits > 7).mean() * 100

print(pct_waiting_over_7min()) 
```

```python
# 10 
# simple growth simulation

def pct_waiting_over_7min(n_customers=10000, seed=42):
    
    np.random.seed(seed)
    waits = np.random.normal(5, 1.5, size=n_customers)
    return (waits > 7).mean() * 100

print(pct_waiting_over_7min()) 
```

```python
# stretch challenge
def monte_carlo_portfolio(n_simulations=1000, start_value=1_000_000, n_days=252, seed=42):
    
    np.random.seed(seed)
    daily_returns = np.random.normal(0.0005, 0.015, size=(n_simulations, n_days))
    growth_factors = np.cumprod(1 + daily_returns, axis=1)
    final_prices = start_value * growth_factors[:, -1]
    mean_final_price = final_prices.mean()
    pct_below_start = (final_prices < start_value).mean() * 100
    return mean_final_price, pct_below_start

print(monte_carlo_portfolio())
```
